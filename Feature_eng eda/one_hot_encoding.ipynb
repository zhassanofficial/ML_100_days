OneHotEncoding using sklearn

```python
from sklearn.model_selection import train_test_split
```

```python
df.head()
```

```python
x_train , x_test , y_train , y_test = train_test_split(df.drop('selling_price' , axis = 1),df['selling_price'],
                                                       test_size=0.2,random_state=0)
```

```python
y_test.shape
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
oh = OneHotEncoder(drop='first')
```

```python
x_train_encoded = oh.fit_transform(x_train[['fuel','owner']]).toarray()
x_test_encoded = oh.transform(x_test[['fuel','owner']]).toarray()
```

```python
x_train_encoded
```

```python
x_test_encoded
```

```python
x_train.head()
```

```python
new_df = np.hstack((x_train[['brand','km_driven']].values,x_train_encoded))
```

```python
pd.DataFrame(new_df).sample(10)
```

df['brand'] feature encoding

```python
counts = df['brand'].value_counts()
```

```python
df['brand'].nunique()
threshold = 100
```

```python
repl = counts[counts <= threshold].index
```

```python
pd.get_dummies(df['brand'].replace(repl, 'uncommon')).sample(10)
```
