```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
```

```python
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score

from sklearn.linear_model import LinearRegression

from sklearn.metrics import r2_score

from sklearn.preprocessing import PowerTransformer
```

```python
df = pd.read_csv('concrete_data.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.shape
```

```python
x = df.drop(columns='Strength')
y = df.iloc[:,-1]
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
lr = LinearRegression()
```

```python
#applying regression without any transformation
lr.fit(x_train,y_train)
pred = lr.predict(x_test)
r2_score(y_test,pred)
```

```python
#cross validation check
lr = LinearRegression()
np.mean(cross_val_score(lr,x,y,scoring="r2"))
```

```python

for col in x_train.columns:

    plt.figure(figsize=(12,5))

    # First subplot
    plt.subplot(1,2,1)
    sns.histplot(x_train[col], kde=True)
    plt.title(f"{col} Distribution")

    # Second subplot
    plt.subplot(1,2,2)
    stats.probplot(x_train[col], dist="norm", plot=plt)
    plt.title(f"{col} QQ Plot")

    plt.tight_layout()
    plt.show()
```

```python
#applying box cox transformation
pt = PowerTransformer(method='box-cox')
x_train_transformed = pt.fit_transform(x_train+0.000001)
x_test_transformed = pt.transform(x_test+0.000001)
pd.DataFrame({'cols':x_train.columns,'box_cos_landa':pt.lambdas_})
```

```python
#applying linear regression on transformed 
lr = LinearRegression()
lr.fit(x_train_transformed,y_train)
y_pred = lr.predict(x_test_transformed)
r2_score(y_test,y_pred)
```

```python
#using cross val score
pt = PowerTransformer(method='box-cox')

x_transformed = pt.fit_transform(x+0.0001)
lr = LinearRegression()

np.mean(cross_val_score(lr,x_transformed,y,scoring='r2'))
```

```python
#applying yeo jonson transformation
pt = PowerTransformer()
x_train_transformed = pt.fit_transform(x_train+0.000001)
x_test_transformed = pt.transform(x_test+0.000001)
print(pd.DataFrame({'cols':x_train.columns,'yeo_jonson_lamda':pt.lambdas_}))

lr = LinearRegression()
lr.fit(x_train_transformed,y_train)
y_pred = lr.predict(x_test_transformed)
print(r2_score(y_test,y_pred))
```
