```python
import numpy as np
import pandas as pd 
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'Product_Code': [
        '2c',
        '3a',
        '5b',
        '123xy',
        '2a5b',
        np.nan,
        '7z',
        '10ab',
        '4x',
        '99yz',
        np.nan,
        '15m',
        '8n',
        '200pq',
        '6k'
    ]
})
df
```

```python
#filling the missing values
si = SimpleImputer(strategy='most_frequent')
df[['Product_Code']] = si.fit_transform(df[['Product_Code']])
```

```python
#extracting alphabetics
df['Number'] = pd.to_numeric(
    df['Product_Code'].str.extract(r'(\d+)')[0],
    errors='coerce'
)

df['Letters'] = df['Product_Code'].str.extract(r'([A-Za-z]+)')
```
