```python
import numpy as np
import pandas as pd
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.read_csv('covid_toy.csv')
df.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test = train_test_split(df.drop(columns=['has_covid']),
                                                       df['has_covid'],test_size=0.2)
```

```python
#adding simple imputer to fever class
si = SimpleImputer(strategy='most_frequent')

x_train_fever = si.fit_transform(x_train[['fever']])
x_test_fever = si.fit_transform(x_test[['fever']])
```

```python
x_train_fever.shape
```

```python
# ordinal_encoding -> cough

oe = OrdinalEncoder(categories=[['Mild','Strong']])

x_train_cough = oe.fit_transform(x_train[['cough']])

x_test_cough = oe.transform(x_test[['cough']])
```

```python
x_train.head()
```

```python
#OneHotEncoder --> gender , city
oe = OneHotEncoder(drop='first',sparse_output=False)

x_train_gender_city = oe.fit_transform(x_train[['gender','city']])

x_test_gender_city = oe.fit_transform(x_test[['gender','city']])
```

```python
x_train_gender_city.shape
```

```python
x_train.head(2)
```

```python
#Extracting age column
x_train_age = x_train.drop(columns=['gender','fever','cough','city'])

x_test_age = x_test.drop(columns=['gender','fever','cough','city'])
```

```python
#concatination
x_trian_transformed = np.concatenate((x_train_age,x_train_cough,x_train_fever,x_train_gender_city),axis=1)

x_test_transformed = np.concatenate((x_test_age,x_test_cough,x_test_fever,x_test_gender_city),axis=1)
```
