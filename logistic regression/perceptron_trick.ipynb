```python
import numpy as np
from sklearn.datasets import make_classification
```

```python
x , y = make_classification(n_samples= 100  ,n_features=2, n_classes=2 ,hypercube= False,class_sep=30,
                            n_redundant=0 , n_informative=1, n_clusters_per_class=1,random_state=41)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,4))
plt.scatter(x[:,0],x[:,1],c=y , cmap='winter' , s=100)
```

```python
def perceptron(x,y):
    x = np.insert(x,0,1,axis=1)
    weights = np.ones(x.shape[1])
    lr = 0.01
    for i in range(2000):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(x[j],weights))
        weights = weights + lr * (y[j] - y_hat) * x[j]
        return weights[0] , weights[1:]
```

```python
intercept_ , coef_ = perceptron(x,y)
print(intercept_)
print(coef_)
```

```python
def step(z):
    return 1 if z>0 else 0
```

Drawing m and b

```python
m = - (coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input = np.linspace(-3,3,100)
y_input = m * x_input + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color = 'red' , linewidth = 3)
plt.scatter(x[:,0],x[:,1],c=y , cmap='winter' , s= 100)
plt.ylim(-3,2)
```

Sigmoid function

```python
def Seg(x,y):
    x = np.insert(x,0,1,axis=1)
    weights = np.ones(x.shape[1])
    lr = 0.01
    for i in range(1000):
        j = np.random.randint(0,100)
        y_hat = segmoid(np.dot(x[j],weights))
        weights = weights + lr * (y[j] - y_hat) * x[j]
        return weights[0] , weights[1:]
```

```python
def segmoid(z):
    return 1/(1+np.exp(-z) )
```

```python
intercept_ , coef_ = Seg(x,y)
print(intercept_)
print(coef_)
```

```python
m = - (coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
input = np.linspace(-3,3,100)
input = m * x_input + b
```
