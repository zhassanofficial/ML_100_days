```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('train.csv',usecols=['GarageQual','FireplaceQu','SalePrice'])
```

```python
df.columns
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
df['GarageQual'].value_counts().plot(kind='bar')
```

```python
fig = plt.figure(figsize=(6,4))
ax = fig.add_subplot(111)
df[df['GarageQual'] == 'TA']['SalePrice'].plot(kind = 'kde',ax = ax,label='GarageQual = TA')
df[df['GarageQual'].isnull()]['SalePrice'].plot(kind = 'kde',ax = ax,color = 'red',label='GarageQual = NAN')
plt.legend()
plt.show()
```

```python
temp = df[df['GarageQual'] == 'TA']['SalePrice']
```

```python
df['GarageQual'].fillna('TA' ,  inplace=True)
```

```python
fig = plt.figure(figsize=(6,4))
ax = fig.add_subplot(111)
temp.plot(kind='kde',ax=ax)
df[df['GarageQual'] == 'TA']['SalePrice'].plot(kind = 'kde',ax = ax,label='GarageQual = TA',color ='blue')
plt.legend()
plt.show()
```
